```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Create some random data to plot
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)
```

```python
# Make a basic scatterplot
fig,ax = plt.subplots(figsize=(5,3)) #create two objects, a figure object and axis object, which is how matplotlib create figures
ax.scatter(x,y) # Then we can call the type of plot we want
```

```python
# Make a barplot
fig,ax = plt.subplots(figsize=(5,3)) 
ax.bar(x,y)
```

```python
# Make a line plot
fig,ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y)
```

```python
# Make a histogram
fig,ax = plt.subplots(figsize=(5,3)) 
ax.hist(y, edgecolor = 'black') # Use y only when doing histogram since there is no x axis
```

```python
# Adding titles and labels
fig,ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout() # remove unecessary blank space
```

```python
# modifying fonts & text color
font1 = {'family' : 'sans-serif', 'color' : 'blue', 'size': 20}
font2 = {'family' : 'monospace', 'color' : 'green', 'size': 14}
```

```python
fig,ax = plt.subplots(figsize=(5,3)) 
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
# Modifying data points: type of points and color
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
           y,
          marker='*',         # NEW ADDITIONS
          color = "indigo")   # NEW ADDITIONS
plt.show()
```

```python
# More modification: line style and width
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='*',
        color = 'indigo',
        linestyle = '--',
        linewidth = 2)
fig.show()
# NEW ADDITIONS
# NEW ADDITIONS
```

```python
# Color: can use HEX codes too
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',
        markersize = 12,              # NEW ADDITIONS
        color = "#6B2FA0",
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',  # NEW ADDITIONS
        markerfacecolor = "#D04040" ) # NEW ADDITIONS
plt.show()
```

```python
# Adding grid lines
fig, ax = plt.subplots(figsize=(5, 3))
ax.grid(axis = 'y') # simple line grid
ax.plot(x,y,marker='*',
        markersize = 12,              # NEW ADDITIONS
        color = "#6B2FA0",
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',  # NEW ADDITIONS
        markerfacecolor = "#D04040" ) # NEW ADDITIONS
plt.show()
```

```python
# Modifying grid lines
fig, ax = plt.subplots(figsize=(5, 3))
ax.grid(axis = 'y', color = "blue", linewidth = 2, linestyle = '-.')
ax.plot(x,y,marker='*',
        markersize = 12,              # NEW ADDITIONS
        color = "#6B2FA0",
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',  # NEW ADDITIONS
        markerfacecolor = "#D04040" ) # NEW ADDITIONS
plt.show()
```

```python
# Recreate a plot from https://python-graph-gallery.com/all-charts/
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('iris')
```

```python
#Simple density plot
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length)
plt.show()
```

```python
#Add a colormap red
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Reds", fill=True)
plt.show()
```

```python
# Change smoothing parameters with bw_adjust
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length,
            cmap="Reds", fill=True, bw_adjust=0.5)
plt.show()
```
