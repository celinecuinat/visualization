### Subplots

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig,(ax1, ax2) = plt.subplots(ncols=2,
                        nrows=1,
                        figsize=(7, 3)) # specify the size of the two plots

ax1.scatter(x1,y1)
ax1.annotate('This is important', xy = (10, 65), xytext=(30, 80), arrowprops = dict(arrowstyle = "wedge", color = "black"))
ax2.bar(x2,y2)

```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], #layout for the 1st row
                                ['ax2', 'ax3']], # layout for the 2nd row
                                 figsize = (7,4)) 
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

```

```python
#Layouts

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                ['ax2', 'ax3']],
                                figsize=(7, 4),
                                layout = "constrained") # constrained layout here
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )

# fig.tight_layout() #or tight layout here

# To adjust the layout and avoid text being cut and so, we can use tight layout or constrained layout
```

```python
# Multiple visualization on one axes
#first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
# Add error information
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

ax.errorbar(x, #our x values
    y2, #our y values
    yerr = y2_sd,
    fmt = "none")
```

```python
#modify errorbars

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
    color = "red")
ax.errorbar(x,
    y2,
    yerr = y2_sd,
    fmt = "none",
    ecolor= "indigo",
    elinewidth= 4,
    capsize = 6,
    capthick= 4 )
```

```python
#modify errorbars intervals (if not necessary for each points)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
    color = "red")

ax.errorbar(x,
        y2,
        yerr = y2_sd,
        fmt = "none",
        ecolor = "blue",
        elinewidth = 4,
        capsize = 6,
        capthick = 4,
        errorevery= 2 )
```

```python
# Using images from the internet
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"}
response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
x=np.array(['Luffy','Zoro','Nami','Usopp','Sanji'])
y1=np.array([110,180,240,99,220])
y2=np.array([170,100,90,120,50])
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='red')
ax_image=fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                       0.11, # y coordinate (ON FIGURE, NOT AXES)
                       0.15, # image width
                       0.35]) # image height)
ax_image.imshow(image)
ax_image.axis('off')
```
