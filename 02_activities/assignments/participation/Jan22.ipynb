```python
## Import libraries
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613) #Set the seed
x = np.arange(50) #array from 0 to 49
y1 = np.random.randint(0, 100,50) #0 to 100 with 50 random integers
y2 = np.random.randint(0, 100,50) #0 to 100 with 50 random integers
```

### Plotting two variables on the same graph

```python
# Line plot with two variables
fig,ax = plt.subplots(figsize=(5,3)) # create matplotlib object
ax.plot(x,y1) # use plot to do a line graph, could use scatter, hist,...
ax.plot(x,y2)
# or use directly: 
# ax.plot(x,y1, y2)
#fig.show() # Not necessary with VScode, figure appears automatically.
```

### Adding plot legends

```python
# Same code but adding variables legend within the plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='lower right') 

# or "best"
# "upper right"
# "upper left"
# "lower left"
# "lower right"
# "right"
# "center left"
# "center right"
# "lower center"
# "upper center"
# "center"
```

```python
# More option to edit the legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
        frameon = True, #add frame around the legend
        fontsize = 12, #change font size
        ncol = 2, #specify number of columns instead of one above
        shadow = True )
```

```python
# Adding variables legend outside the plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',
        bbox_to_anchor =(1, 1)) # could do upper left with (1,0)


# (0,0) is the origin of the graph (bottom left)
# (1,1) represent the top right corner
```

### Annotate a plot with a text or arrow

```python
# Scatter plot and adding annotations using coordinates
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!") # value 10 on x-axis and 95 in y axis
```

```python
# Modify text annotation
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!",
      ha='center',    ##Horizontal alignment of text
      color = 'red',  ##modify font colour
      size = 20)      ##modify font size
```

```python
# Adding annotations to a fixed spot
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1, 5)",           transform=ax.transData    ) # annotation using our data as reference
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)",   transform=ax.transAxes    ) # using the axes
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure ) # using the figure dimensions
```

```python
# Using arrows
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
          arrowprops=dict(facecolor='black'))
```

```python
# Modifying arrows
# https://matplotlib.org/3.1.1/api/_as_gen/matplotlib.pyplot.annotate.html
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94), arrowprops = dict(arrowstyle = "wedge", color = "hotpink"))
```

### Modifying axis labels

```python
# removing tick marks or labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator()) # NullLocator for nothing
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
# Limiting the number of tick marks used
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MaxNLocator(3)) # MaxNLocator: use only x tick marks
```

```python
# Changing tick mark intervals
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #MultipleLocator: set interval between marks
```

```python
# Rotating axis labels direction
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
          
plt.xticks(rotation=45, ha='right')
style = {"family": "serif", "size": 30, "color": "Indigo"} # Create a style to apply to the label
plt.xlabel("New X Axis!", fontdict= style)

# or plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
```
